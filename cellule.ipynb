```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.linalg as la

def convergence_kalman_riccati(v1, temps, R_circuit=1.0, L_circuit=0.1, C_circuit=0.01, 
                               sigma_modele=0.1, R_mesure=2.083):
    

    v1 = np.array(v1)
    temps = np.array(temps)
    n_steps = len(v1)
    dt = temps[1] - temps[0] if n_steps > 1 else 0.001
    
    A = np.array([[1.0, dt / C_circuit],
                  [-dt / L_circuit, 1.0 - (R_circuit * dt) / L_circuit]])
    H = np.array([[1.0, 0.0]])
    
    Phi = np.array([[0.0, 0.0], 
                    [0.0, ((dt / L_circuit)**2) * (sigma_modele**2)]])
    Psi = np.array([[R_mesure]])

    
    P_moins_inf = la.solve_discrete_are(A.T, H.T, Phi, Psi)
    
    S_inf = H @ P_moins_inf @ H.T + Psi
    K_inf = P_moins_inf @ H.T @ np.linalg.inv(S_inf)
    P_inf = (np.eye(2) - K_inf @ H) @ P_moins_inf

    P = np.eye(2) * 10.0  
    
    K_historique = []
    erreur_P_historique = []
    
    for n in range(1, n_steps):
        P_pred = A @ P @ A.T + Phi
        
        S = H @ P_pred @ H.T + Psi
        K = P_pred @ H.T @ np.linalg.inv(S)
        
        P = (np.eye(2) - K @ H) @ P_pred
        
        K_historique.append(K.flatten())
        erreur_P_historique.append(np.linalg.norm(P - P_inf, 'fro'))

    K_historique = np.array(K_historique)

    print("\n" + "="*50)
    print("CONVERGENCE DES MATRICES DE KALMAN")
    print("="*50)
    print("\n--- Solution de l'Équation de Riccati ---")
    print(f"P_inf (Théorique) =\n{P_inf}")
    print(f"K_inf (Théorique) =\n{K_inf}")
    
    print("\n--- Solution Itérative (à la fin de la liste) ---")
    print(f"P_final (Itératif) =\n{P}")
    print(f"K_final (Itératif) =\n{K}")
    
    diff_P = np.linalg.norm(P - P_inf)
    print(f"\nDifférence absolue (Norme) : {diff_P:.2e}")
    if diff_P < 1e-5:
        print("-> EXCELLENT : L'algorithme a parfaitement convergé !")
    
    temps_plot = temps[1:]
    fig, axes = plt.subplots(2, 1, figsize=(10, 8))
    
    axes[0].plot(temps_plot, K_historique[:, 0], 'c-', linewidth=2, label='K[0] (Tension) - Itératif')
    axes[0].plot(temps_plot, K_historique[:, 1], 'm-', linewidth=2, label='K[1] (Courant) - Itératif')
    axes[0].axhline(K_inf[0,0], color='blue', linestyle='--', label='K[0] Riccati')
    axes[0].axhline(K_inf[1,0], color='red', linestyle='--', label='K[1] Riccati')
    axes[0].set_title("Convergence du Gain de Kalman vers le régime permanent", fontweight='bold')
    axes[0].set_ylabel("Valeur du Gain")
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)
    
    axes[1].plot(temps_plot, erreur_P_historique, 'r-', linewidth=2)
    axes[1].set_yscale('log')
    axes[1].set_title(r"Vitesse de convergence : Norme de l'erreur $|| P_n - P_{\infty} ||$", fontweight='bold')
    axes[1].set_xlabel("Temps (s)")
    axes[1].set_ylabel("Erreur (Log scale)")
    axes[1].grid(True, which="both", alpha=0.3)
    
    plt.tight_layout()
    plt.show()

convergence_kalman_riccati(v1, temps)
```
